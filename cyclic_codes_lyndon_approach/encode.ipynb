```python
filename = 'lyndon_words/elf37_cycle_reps.txt'
count, unique_roots = count_root_codewords_from_grid(filename)
print(f"Number of root messages: {count}")

n_rows = len(unique_roots)
n_cols = 15

vv_msgs = np.empty((n_rows, n_cols), dtype=np.float32)
period = np.empty((n_rows,))

for i, v_msg in enumerate(unique_roots):
  vv_msgs[i] = np.array(v_msg, dtype=np.float32)
  period[i] = find_smallest_repeat(v_msg)


print("weight: ", np.sum(vv_msgs, axis=1))
print("period: ", period)
```

```python
info = tbcc_encoder.CodeInformation(1, 2, 6, 0, 0, 15, [133, 171])

# 2. Instantiate the Trellis class
trellis = tbcc_encoder.FeedForwardTrellis(info)

nextStates = trellis.getNextStates()
output = trellis.getOutputs()

print("nextStates: ", nextStates)
print("outputs: ", output)
```

```python
# encoding
# 1. Setup the CodeInformation (using your constructor's arguments)
# Example values: k, n, v, crcLen, crc, numInfoBits, numerators_vector
info = tbcc_encoder.CodeInformation(1, 2, 5, 0, 0, 15, [43, 57])

# 2. Instantiate the Trellis class
trellis = tbcc_encoder.FeedForwardTrellis(info)

# 3. Call encode ON THE INSTANCE
# Note: Ensure your C++ 'encode' accepts the type you are passing
spectrum = np.zeros(shape=(31,))
min_neighbor_lyndon_words = []
for i, v_msg in enumerate(vv_msgs):
    cwd = trellis.encode(v_msg)
    weight = np.sum(cwd)
    spectrum[weight] += period[i]
    if weight == 10:
        min_neighbor_lyndon_words.append(v_msg)
        print("----")
        print(f"v_msg: {' '.join(str(int(x)) for x in v_msg)}")
        print("period: ", period[i])
        print("cwd: ", {' '.join(str(int(x)) for x in cwd)})
        c1 = cwd[0::2]
        c2 = cwd[1::2]
        print("c1: ", {' '.join(str(int(x)) for x in c1)})
        print("c2: ", {' '.join(str(int(x)) for x in c2)})
        c1_spc = np.mod(np.sum(c1),2)
        c2_spc = np.mod(np.sum(c2),2)
        print("c1_spc: ", c1_spc)
        print("c2_spc: ", c2_spc)

print("spectrum: ", spectrum)
```
